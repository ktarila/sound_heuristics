```python
import numpy as np
import math
import matplotlib.pyplot as plt
```

```python
x_array = np.linspace(-500, 500, 1000)
print(x_array)
```

```python
def f(x):
    # function decays to 37% of previous value at
    # every 100 intervals. Initial value at x = 0 is 90
    # return 90 * math.exp(-1 * x/100)
    return np.sum((-1 * x) * np.sin(np.sqrt(np.abs(x))))
```

```python
f(0)
```

```python
y_array = [f(x) for x in x_array]
```
